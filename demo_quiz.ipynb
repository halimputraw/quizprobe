Define a function sort_inventory(information) which:
- Takes one string containing multiple inventory records, for example:
  "Pencil, 12, 3; Notebook, 8, 5".
- Each record is separated by a semicolon (;).
- Each record contains three comma-separated fields:
  - item name (str)
  - quantity (int)
  - unit price (int)
- Returns a list of tuples in the form (name, quantity, unit_price).
- Sorts the returned list in ascending lexicographical order by name. If names are identical, sort by quantity and then unit price.

Hint: Remove leading and trailing whitespace from each item name and from the numeric fields before converting the numbers to integers.

```python

def sort_inventory(information):
    result = []

    for record in information.split(';'):
        name, quantity, price = record.split(',')

        name = name.strip()
        quantity = int(quantity.strip())
        price = int(price.strip())

        result.append((name, quantity, price))

    result.sort()
    return result
```
